```python
# Check label conflicts 
from sklearn.model_selection import train_test_split

X = dataset["document_text"]
y = dataset["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20, #means 20% is held aside for testing and 80% is used for training.
    random_state=42, #gives you a reproducible split. Running the notebook again gives you the same train/test allocation.
    stratify=y
)

print("TRAIN")
print(y_train.value_counts())

print("\nTEST:")
print(y_test.value_counts())

```

```python
#build the baseline text classifier.
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1,2),
    min_df=2
)

X_train_vec = vectorizer.fit_transform(X_train)
x_test_vec = vectorizer.transform(X_test)

print(X_train.shape)
print(x_test_vec.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000, #means: Logistic Regression doesn't instantly know the best 4 × 9097 weights.
    # An optimizer gradually adjusts them: initial weights -> calculate errors -> adjust coefficients -> calculate errors again -> adjust
    class_weight="balanced"
)

model.fit(X_train_vec, y_train)
```

```python

from sklearn.metrics import classification_report

#after training generate predictions
y_pred = model.predict(x_test_vec)

#then evaluate
print(
    #is the evaluation step. It compares: y_test correct answers and y_pred model's answers 
    classification_report( 
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
#STRATIFIED GROUP K-FOLD CROSS-VALIDATION 
from sklearn.model_selection import StratifiedGroupKFold

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

groups = dataset["document_text"]

#inspect the folds before training anything
#why 5 folds - enough repetitions for a better performance estimate and not too computationally expensive
for fold, (train_idx, test_idx) in enumerate(
    cv.split(dataset["document_text"], dataset["label"], groups)
):
    print(f"\nFold {fold + 1}")
    print("TRAIN:")
    print(dataset.iloc[train_idx]["label"].value_counts())
    print("TEST:")
    print(dataset.iloc[test_idx]["label"].value_counts())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_validate

lr_pipeline = Pipeline([ #Pipeline is an ML pipeline. It chains preprocessing and modelling together
    ("tfidf", TfidfVectorizer(
        lowercase=True, ngram_range=(1,2), min_df=2
    )),
    ("model", LogisticRegression(
        max_iter=1000, class_weight="balanced"
    ))
])

#evaluate
scores = cross_validate(
    lr_pipeline,
    dataset["document_text"],
    dataset["label"],
    groups=groups,
    cv=cv,
    scoring=["accuracy", "f1_macro"]
)

print("Accuracy:", scores["test_accuracy"])
print("Macro F1:", scores["test_f1_macro"])

print("Mean Accuracy:", scores["test_accuracy"].mean())
print("Mean Macro F1:", scores["test_f1_macro"].mean())
```

```python
#5-fold Stratified Group K-Fold cross-validation for Linear SVM
#so we can compare it fairly against logistic regression

svm_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True, ngram_range=(1, 2), min_df=2
    )),
    ("model", LinearSVC(
        class_weight="balanced"
    ))
])

svm_scores = cross_validate(
    svm_pipeline,
    dataset["document_text"],
    dataset["label"],
    groups=groups,
    cv=cv,
    scoring=["accuracy", "f1_macro"]
)


#inspect
print("Accuracy:", svm_scores["test_accuracy"])
print("Macro F1:", svm_scores["test_f1_macro"])
print("Mean Accuracy:", svm_scores["test_accuracy"].mean())
print("Mean Macro F1:", svm_scores["test_f1_macro"].mean())
```

```python
#use cross_val_predict() so every row receives a predictions from a model that did not train on that row
from sklearn.model_selection import cross_val_predict

svm_cv_pred=cross_val_predict(
    svm_pipeline,
    dataset["document_text"],
    dataset["label"],
    groups=groups,
    cv=cv,
)

print(classification_report(
    dataset["label"] ,
    svm_cv_pred,
    zero_division=0
))
```
